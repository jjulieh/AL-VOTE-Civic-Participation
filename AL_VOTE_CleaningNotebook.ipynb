**#3: Needed Function**


*   Removes commas/spaces


*   Converts values to whole numbers
*   Blank values become missing values





```python
def clean_numeric_column(series):
    cleaned = (
        series.astype("string")
        .str.strip()
        .str.replace(",", "", regex=False)
        .str.replace(r"\s+", "", regex=True)
        .replace({
            "": pd.NA,
            "NA": pd.NA,
            "N/A": pd.NA,
            "nan": pd.NA,
        })
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    ).astype("Int64")
```

**#4: Cleaning**

*   Renames columns to be easier for SQL
* Cleans county names

*   Removes the statwide totals row

*   Adds year as the first column


*   Identifies all numeric columns
*   Removes commas and converts numeric columns to integers


*   Keeps only the 43 relevant counties
*   Sorts alphabetically








```python
age_2020 = age_2020_raw.copy()

age_2020 = age_2020.rename(
    columns={
        "County": "county_name",
        "Total Ballots Cast": "total_participants",
        "Age 18-29": "age_18_29",
        "Age 30-39": "age_30_39",
        "Age 40-49": "age_40_49",
        "Age 50-59": "age_50_59",
        "Age 60-69": "age_60_69",
        "Age 70-79": "age_70_79",
        "Age 80-89": "age_80_89",
        "Age 90-99": "age_90_99",
        "Age 100+": "age_100_plus",
    }
)

age_2020["county_name"] = (
    age_2020["county_name"]
    .astype("string")
    .str.strip()
    .str.replace("_", " ", regex=False)
    .str.lower()
    .str.title()
)

age_2020["county_name"] = age_2020["county_name"].replace(
    {
        "Dekalb": "DeKalb",
        "St Clair": "St. Clair",
    }
)

age_2020 = age_2020[
    age_2020["county_name"].str.upper() != "TOTALS"
].copy()

age_2020.insert(0, "year", 2020)

numeric_columns = [
    "total_participants",
    "age_18_29",
    "age_30_39",
    "age_40_49",
    "age_50_59",
    "age_60_69",
    "age_70_79",
    "age_80_89",
    "age_90_99",
    "age_100_plus",
]

for column in numeric_columns:
    age_2020[column] = clean_numeric_column(
        age_2020[column]
    )

age_2020 = age_2020[
    age_2020["county_name"].isin(PROJECT_COUNTIES)
].copy()

age_2020 = age_2020.sort_values(
    "county_name"
).reset_index(drop=True)

print("Cleaned shape:", age_2020.shape)

age_2020.head()
```



---



**#5: Checking**
 if we have data for all 43 counties

*   Data for all 43 counties
*   Duplicates or missing values


*   If age groups equal total population



```python
counties_found = set(age_2020["county_name"])
counties_expected = set(PROJECT_COUNTIES)

missing_counties = sorted(
    counties_expected - counties_found
)

unexpected_counties = sorted(
    counties_found - counties_expected
)

print("Missing counties:", missing_counties)
print("Unexpected counties:", unexpected_counties)
```

```python
duplicate_rows = age_2020[
    age_2020.duplicated(
        subset=["year", "county_name"],
        keep=False
    )
]

print("Number of duplicate rows:", len(duplicate_rows))

duplicate_rows
```

```python
missing_values = age_2020.isna().sum()

print(missing_values)
```

```python
age_group_columns = [
    "age_18_29",
    "age_30_39",
    "age_40_49",
    "age_50_59",
    "age_60_69",
    "age_70_79",
    "age_80_89",
    "age_90_99",
    "age_100_plus",
]

age_2020["calculated_age_total"] = (
    age_2020[age_group_columns].sum(axis=1)
)

age_2020["age_total_difference"] = (
    age_2020["calculated_age_total"]
    - age_2020["total_participants"]
)

age_discrepancies = age_2020.loc[
    age_2020["age_total_difference"] != 0,
    [
        "county_name",
        "total_participants",
        "calculated_age_total",
        "age_total_difference",
    ]
]

print(
    "Counties with age-total discrepancies:",
    len(age_discrepancies)
)

age_discrepancies
```

```python
age_2020 = age_2020.drop(
    columns=[
        "calculated_age_total",
        "age_total_difference",
    ]
)

print(age_2020.columns.tolist())
```

**Needed county name function**


*   extracts and standardizes the county name from the files
* removes the year and general
* removes spaces
* standardizes the capitialization
* corrects the special names






```python
def extract_county_from_filename(file_path):

    county_name = file_path.stem

    county_name = re.sub(
        r"^\d{4}[-_\s]+General[-_\s]+",
        "",
        county_name,
        flags=re.IGNORECASE
    )

    county_name = re.sub(
        r"\(\d+\)$",
        "",
        county_name
    )

    county_name = county_name.replace("_", " ")
    county_name = county_name.replace("-", " ")


    county_name = re.sub(
        r"\s+",
        " ",
        county_name
    ).strip()

    county_name = county_name.lower().title()

    county_name = {
        "Dekalb": "DeKalb",
        "St Clair": "St. Clair",
        "St. Clair": "St. Clair",
        "Stclair": "St. Clair",
    }.get(
        county_name,
        county_name
    )

    return county_name
```

**Importer Function**


*   extrats the country name
* identifies precinct and voting method columns
* converts into numeric values
* sums into total votes
* standardizes



```python
def import_county_election_file(
    file_path,
    year
):

    df = pd.read_excel(
        file_path,
        engine="xlrd"
    )

    county_name = (
        extract_county_from_filename(
            file_path
        )
    )

    required_columns = [
        "Contest Title",
        "Party",
        "Candidate",
    ]

    missing_columns = [
        column
        for column in required_columns
        if column not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"{file_path.name} is missing "
            f"required columns: {missing_columns}"
        )
    vote_columns = [
        column
        for column in df.columns
        if column not in required_columns
    ]

    for column in vote_columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

    df["total_votes"] = (
        df[vote_columns]
        .sum(
            axis=1,
            min_count=1
        )
    )

    df.insert(
        0,
        "year",
        year
    )

    df.insert(
        1,
        "county_name",
        county_name
    )

    df["source_file"] = file_path.name

    df = df[
        [
            "year",
            "county_name",
            "Contest Title",
            "Party",
            "Candidate",
            "total_votes",
            "source_file",
        ]
    ].copy()

    df = df.rename(
        columns={
            "Contest Title": "contest_title",
            "Party": "party",
            "Candidate": "candidate",
        }
    )

    return df
```

Text Cleaning function

```python
def clean_election_text(series):
    return (
        series
        .astype("string")
        .str.strip()
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
        .replace(
            {
                "": pd.NA,
                "nan": pd.NA,
                "<NA>": pd.NA,
            }
        )
    )
```

Election Cleaning Function:


*  makes total_votes numberic
*   removes negative totals
* keeps the needed contests
* standardizes the contest type



```python
def clean_combined_election_data(
    combined_raw
):

    df = combined_raw.copy()

    text_columns = [
        "county_name",
        "contest_title",
        "party",
        "candidate",
        "source_file",
    ]

    for column in text_columns:
        df[column] = clean_election_text(
            df[column]
        )

    df["total_votes"] = pd.to_numeric(
        df["total_votes"],
        errors="coerce"
    ).round().astype("Int64")

    df = df[
        df["county_name"].notna()
        & df["contest_title"].notna()
        & df["candidate"].notna()
        & df["total_votes"].notna()
    ].copy()

    df = df[
        df["total_votes"] >= 0
    ].copy()

    presidential_mask = (
        df["contest_title"]
        == "PRESIDENT AND VICE PRESIDENT OF THE UNITED STATES"
    )

    congressional_mask = (
        df["contest_title"]
        .str.startswith(
            "UNITED STATES REPRESENTATIVE",
            na=False
        )
    )

    df = df[
        presidential_mask
        | congressional_mask
    ].copy()

    df["contest_type"] = pd.NA

    df.loc[
        df["contest_title"]
        == "PRESIDENT AND VICE PRESIDENT OF THE UNITED STATES",
        "contest_type"
    ] = "president"

    df.loc[
        df["contest_title"]
        .str.startswith(
            "UNITED STATES REPRESENTATIVE",
            na=False
        ),
        "contest_type"
    ] = "us_house"

    df["district_number"] = (
        df["contest_title"]
        .str.extract(
            r"(\d+)(?:ST|ND|RD|TH)\s+CONGRESSIONAL",
            expand=False
        )
    )

    df["district_number"] = pd.to_numeric(
        df["district_number"],
        errors="coerce"
    ).astype("Int64")

    df.loc[
        df["contest_type"] == "president",
        "district_number"
    ] = pd.NA

    candidate_lower = (
        df["candidate"]
        .astype("string")
        .str.lower()
    )

    df["result_type"] = "candidate"

    df.loc[
        candidate_lower == "write-in",
        "result_type"
    ] = "write_in"

    df.loc[
        candidate_lower == "over votes",
        "result_type"
    ] = "overvote"

    df.loc[
        candidate_lower == "under votes",
        "result_type"
    ] = "undervote"

    df.loc[
        df["result_type"].isin(
            ["overvote", "undervote"]
        ),
        "party"
    ] = pd.NA

    df = df[
        [
            "year",
            "county_name",
            "contest_type",
            "district_number",
            "contest_title",
            "result_type",
            "party",
            "candidate",
            "total_votes",
            "source_file",
        ]
    ].copy()

    result_order = {
        "candidate": 1,
        "write_in": 2,
        "overvote": 3,
        "undervote": 4,
    }

    df["_result_order"] = (
        df["result_type"]
        .map(result_order)
    )

    df = (
        df
        .sort_values(
            [
                "county_name",
                "contest_type",
                "district_number",
                "_result_order",
                "candidate",
            ],
            na_position="first"
        )
        .drop(
            columns="_result_order"
        )
        .reset_index(drop=True)
    )

    return df
```

**CONTEST SUMMARY FUNCTION**


*   Creates one row per county contest that has canidate votes, overvotes, undervotes, and total participation



```python
def create_election_contest_summary(
    election_clean
):

    summary_source = (
        election_clean.copy()
    )

    summary_source[
        "candidate_votes"
    ] = summary_source[
        "total_votes"
    ].where(
        summary_source[
            "result_type"
        ].isin(
            ["candidate", "write_in"]
        ),
        0
    )

    summary_source[
        "overvotes"
    ] = summary_source[
        "total_votes"
    ].where(
        summary_source[
            "result_type"
        ] == "overvote",
        0
    )

    summary_source[
        "undervotes"
    ] = summary_source[
        "total_votes"
    ].where(
        summary_source[
            "result_type"
        ] == "undervote",
        0
    )

    summary = (
        summary_source
        .groupby(
            [
                "year",
                "county_name",
                "contest_type",
                "district_number",
                "contest_title",
            ],
            dropna=False,
            as_index=False
        )
        .agg(
            candidate_votes=(
                "candidate_votes",
                "sum"
            ),
            overvotes=(
                "overvotes",
                "sum"
            ),
            undervotes=(
                "undervotes",
                "sum"
            ),
            contest_ballots=(
                "total_votes",
                "sum"
            ),
        )
    )

    numeric_columns = [
        "candidate_votes",
        "overvotes",
        "undervotes",
        "contest_ballots",
    ]

    for column in numeric_columns:
        summary[column] = (
            summary[column]
            .astype("Int64")
        )

    return (
        summary
        .sort_values(
            [
                "year",
                "county_name",
                "contest_type",
                "district_number",
            ],
            na_position="first"
        )
        .reset_index(drop=True)
    )
```

```python
election_summary_2020 = (
    create_election_contest_summary(
        election_2020_clean
    )
)

election_summary_2024 = (
    create_election_contest_summary(
        election_2024_clean
    )
)

print(
    "2020 summary shape:",
    election_summary_2020.shape
)

print(
    "2024 summary shape:",
    election_summary_2024.shape
)

election_summary_2020.head(10)
```

This validates presidential totals against cast ballots

```python
def extract_county_ballot_totals(
    combined_raw
):

    df = combined_raw.copy()

    df["contest_title"] = clean_election_text(
        df["contest_title"]
    )

    df["candidate"] = clean_election_text(
        df["candidate"]
    )

    df["total_votes"] = pd.to_numeric(
        df["total_votes"],
        errors="coerce"
    ).round().astype("Int64")

    ballot_totals = df[
        df["contest_title"]
        == "BALLOTS CAST - TOTAL"
    ].copy()

    ballot_totals = ballot_totals[
        [
            "year",
            "county_name",
            "total_votes",
        ]
    ].rename(
        columns={
            "total_votes":
            "official_ballots_cast"
        }
    )

    return (
        ballot_totals
        .drop_duplicates(
            subset=[
                "year",
                "county_name",
            ]
        )
        .sort_values(
            [
                "year",
                "county_name",
            ]
        )
        .reset_index(drop=True)
    )
```

```python
ballot_totals_2020 = (
    extract_county_ballot_totals(
        election_2020_combined_raw
    )
)

ballot_totals_2024 = (
    extract_county_ballot_totals(
        election_2024_combined_raw
    )
)

print(
    "2020 ballot-total rows:",
    len(ballot_totals_2020)
)

print(
    "2024 ballot-total rows:",
    len(ballot_totals_2024)
)
```

```python
president_summary_2020 = (
    election_summary_2020[
        election_summary_2020[
            "contest_type"
        ] == "president"
    ][
        [
            "year",
            "county_name",
            "candidate_votes",
            "overvotes",
            "undervotes",
            "contest_ballots",
        ]
    ]
)

president_validation_2020 = (
    president_summary_2020
    .merge(
        ballot_totals_2020,
        on=[
            "year",
            "county_name",
        ],
        how="outer",
        validate="one_to_one"
    )
)

president_validation_2020[
    "difference"
] = (
    president_validation_2020[
        "contest_ballots"
    ]
    - president_validation_2020[
        "official_ballots_cast"
    ]
)

president_validation_2020
```

```python
president_summary_2024 = (
    election_summary_2024[
        election_summary_2024[
            "contest_type"
        ] == "president"
    ][
        [
            "year",
            "county_name",
            "candidate_votes",
            "overvotes",
            "undervotes",
            "contest_ballots",
        ]
    ]
)

president_validation_2024 = (
    president_summary_2024
    .merge(
        ballot_totals_2024,
        on=[
            "year",
            "county_name",
        ],
        how="outer",
        validate="one_to_one"
    )
)

president_validation_2024[
    "difference"
] = (
    president_validation_2024[
        "contest_ballots"
    ]
    - president_validation_2024[
        "official_ballots_cast"
    ]
)

president_validation_2024
```

```python
president_discrepancies_2020 = (
    president_validation_2020[
        president_validation_2020[
            "difference"
        ].fillna(0) != 0
    ]
)

president_discrepancies_2024 = (
    president_validation_2024[
        president_validation_2024[
            "difference"
        ].fillna(0) != 0
    ]
)

print(
    "2020 presidential discrepancies:",
    len(president_discrepancies_2020)
)

print(
    "2024 presidential discrepancies:",
    len(president_discrepancies_2024)
)
```
